```python
from datetime import datetime

formats = [
    '%A, %B %d, %Y', 
    '%A, %d.%m.%y', 
    '%A, %d %B %Y' 
]


def parse_date(text):
    for date_format in formats:
        try:
            return datetime.strptime(text, date_format)
        except ValueError:
            continue
    return None


while True:
    text = input('Введите дату (q - выход): ').strip()

    if text.lower() == 'q':
        print('Программа завершена.')
        break

    result = parse_date(text)

    if result is None:
        print('Неверный ввод.')
    else:
        print(result)
```
