```python
import csv
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    for line in file:
        purchase = json.loads(line)

        if purchase['user_id'] != 'user_id':
            purchases[purchase['user_id']] = purchase['category']

with open('visit_log__1_.csv', encoding='utf-8', newline='') as visits:
    with open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel:
        reader = csv.DictReader(visits)
        writer = csv.writer(funnel)

        writer.writerow(['user_id', 'source', 'category'])

        for visit in reader:
            user_id = visit['user_id']

            if user_id in purchases:
                writer.writerow([
                    user_id,
                    visit['source'],
                    purchases[user_id]
                ])

with open('funnel.csv', encoding='utf-8') as file:
    for _ in range(3):
        print(file.readline().strip())
```
